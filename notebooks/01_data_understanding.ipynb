```python
import pandas as pd

df = pd.read_csv("../data/creditcard.csv")

print("Dataset shape:", df.shape)
df.head()
```

Time represents the number of seconds elapsed between each transaction and the first transaction in the dataset.

V1 to V28 are transformed/anonymized transaction features. The original banking features were hidden for privacy, and these variables were produced using PCA (Principal Component Analysis). So we don't know whether V1 originally represented location, merchant information, card behaviour, etc.

Amount is the transaction amount.

Class is our target:
    0 = Normal transaction, 
    1 = Fraudulent transaction

```python
print("Columns in the dataset:")
print(df.columns.tolist())
```

```python
print("Missing Values in the dataset:")
print(df.isnull().sum().sum())
```

```python
print("Class distribution:")
print(df["Class"].value_counts()) 
# .value_counts() counts the frequency of each unique value in the "Class" column, 
# which indicates whether a transaction is fraudulent (1) or not (0).
```

Class 0 → 284,315 normal transactions,    
Class 1 → 492 fraud transactions

```python
print("Class percentage:")
print(df["Class"].value_counts(normalize=True) * 100)
```

```python
df.describe()
```

V1–V28 are centered roughly around 0 because they are PCA-transformed features.

```python
df[["Time", "Amount", "Class"]].describe()
```

Amount is highly skewed:

```python
import matplotlib.pyplot as plt

class_counts = df["Class"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(["Normal", "Fraud"], class_counts.values)

plt.title("Class Distribution")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
fraud_percentage = df["Class"].mean() * 100

print(f"Fraud percentage: {fraud_percentage:.4f}%")
print(f"Normal percentage: {100 - fraud_percentage:.4f}%")
```
